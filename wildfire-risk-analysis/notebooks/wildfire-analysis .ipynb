## Data Cleaning

checking for duplicates

```python
import pandas as pd
df = pd.read_csv("../data/forestfires.csv")
df.duplicated().sum()
```

### Duplicate Check

A total of 4 duplicate rows were found in the dataset and removed during cleaning.

```python
# Removing duplicates
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
# Recomfirming For Missing Values

df.isnull().sum()
```

```python
# Checking For Inconsistencies in Categorical Columns
# For Day
df["day"].unique()
```

```python
# For Month
df["month"].unique()
```

```python
# Fixing Month Ordering
month_order = ['jan', 'feb', 'mar', 'apr', 'may', 'jun',
               'jul', 'aug', 'sep', 'oct', 'nov', 'dec']

df["month"] = pd.Categorical(df["month"], categories=month_order, ordered=True)
```

```python
# Fixing Day Ordering
day_order = ['mon', 'tue', 'wed', 'thu', 'fri', 'sat', 'sun']

df["day"] = pd.Categorical(df["day"], categories=day_order, ordered=True)
```

### Categorical Consistency

The categorical columns were checked for inconsistencies using their unique values. Month and day labels were consistent, with no spelling or formatting issues found. Both columns were then converted into ordered categorical variables to preserve their natural calendar order during analysis.

```python
# Inspecting Skewness Of Burnt Area

df["area"].skew()
```

### Skewness in Burned Area

The area column is highly right-skewed, with a skewness value of 12.80. This suggests that most fire incidents affected small areas, while a few cases recorded very large burned areas.
