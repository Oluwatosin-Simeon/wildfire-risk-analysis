### Categorical Consistency

The categorical columns were checked for inconsistencies using their unique values. Month and day labels were consistent, with no spelling or formatting issues found. Both columns were then converted into ordered categorical variables to preserve their natural calendar order during analysis.

## Target Variable Inspection

In this section, the target variable (area), which represents the total burned area of forest fires was inspected. Since this is the variable the model will predict, it is important to understand its distribution, detect skewness, and identify extreme values before modeling.

```python
df['area'].describe()
```

```python
#Visualization of Target Variable
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.histplot(df['area'], bins=30, kde=True)
plt.title('Distribution of Burned Area (Before Transformation)')
plt.xlabel('Burned Area')
plt.ylabel('Frequency')
plt.show()
```

```python
# Checking For Outliers (Extreme Values that deviate significantly from the rest of a dataset in a distribution)
plt.figure(figsize=(8,4))
sns.boxplot(x=df['area'])
plt.title('Boxplot of Burned Area')
plt.show()
```

```python
# Inspecting Skewness Of Burnt Area

df["area"].skew()
```

### Skewness in Burned Area

The area column is highly right-skewed, with a skewness value of 12.80. This suggests that most fire incidents affected small areas, while a few cases recorded very large burned areas.

```python
# Target Variable Log Transformation
import numpy as np

df['area_log'] = np.log1p(df['area'])
```

```python
# Visualizing Transformed Target Variable

plt.figure(figsize=(10,5))
sns.histplot(df['area_log'], bins=30, kde=True)
plt.title('Distribution of Burned Area (After Log Transformation)')
plt.xlabel('Log Transformed Burned Area')
plt.ylabel('Frequency')
plt.show()
```

```python
# Comparison of the before and after of the target Variable
fig, axes = plt.subplots(1, 2, figsize=(14,5))

sns.histplot(df['area'], bins=30, kde=True, ax=axes[0])
axes[0].set_title('Before Transformation')

sns.histplot(df['area_log'], bins=30, kde=True, ax=axes[1])
axes[1].set_title('After Log Transformation')

plt.tight_layout()
plt.show()
```

### Target Variable Treatment

The target variable (area) was highly right-skewed, with many small values and a few extremely large burned-area observations. This created an imbalanced distribution that could negatively affect model learning.

To reduce skewness and minimize the influence of extreme values, a log transformation (log1p) was applied to the target variable.

This transformation improved the distribution of the target variable, making it less skewed and more suitable for deeper exploratory analysis and model development.
